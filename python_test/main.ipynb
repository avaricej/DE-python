```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv')

metrics = pd.pivot_table(df, index='Pclass', columns=['Survived'], values='Age', aggfunc='mean')
difference = metrics[1]-metrics[0]
print(difference)
df.head()

difference.plot(kind='bar')
plt.title('Разница в возрасте')
fig, ax = plt.subplots(figsize=(8,5))
ax.bar(difference.index, difference.values)
```
