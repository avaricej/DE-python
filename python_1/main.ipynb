```python
import csv
import json

purchases=dict()

with open(file='purchase_log.txt', encoding='utf-8', newline='') as mytxt:
    for i, line in enumerate(mytxt):
        if i == 0:
            continue
        # print(type(line))
        ln = json.loads(line)
        
        purchases[ln['user_id']]=ln['category']
    
    
for i, (id, category) in enumerate(purchases.items())   :
    if i>1:
        break;
    print(id,category)
```
